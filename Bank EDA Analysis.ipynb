```python
bins = [0, 100000, 300000, float('inf')]

labels = ['Low', 'Mid', 'High']

df['Income Band'] = pd.cut(
    df['Estimated Income'],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
df['Income Band'].value_counts().plot(kind="bar")
```

```python
# Select categorical columns
categorical_cols = df[
    [
        "BRId",
        "GenderId",
        "IAId",
        "Amount of Credit Cards",
        "Nationality",
        "Occupation",
        "Fee Structure",
        "Loyalty Classification",
        "Properties Owned",
        "Risk Weighting"
    ]
].columns

# Display value counts and bar chart
for col in categorical_cols:
    print(f"Value counts for '{col}':")
    
    display(df[col].value_counts())
    
    df[col].value_counts().plot(
        kind="bar",
        figsize=(8, 4)
    )
    
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
for i, predictor in enumerate(
    df[
        [
            "BRId",
            "GenderId",
            "IAId",
            "Amount of Credit Cards",
            "Nationality",
            "Occupation",
            "Fee Structure",
            "Loyalty Classification",
            "Properties Owned",
            "Risk Weighting"
        ]
    ].columns
):
    
    plt.figure(i, figsize=(8, 4))
    
    sns.countplot(
        data=df,
        x=predictor,
        hue="GenderId"
    )
    
    plt.title(f"{predictor} Distribution by Gender")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Histoplopt of values for differernt occupation

# Create a separate count plot for each categorical column
for col in categorical_cols:
    
    plt.figure(figsize=(8, 4))
    
    sns.countplot(
        data=df,
        x=col
    )
    
    plt.title(f"Count of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

# Neumerical Analysis

```python
numerical_cols = [
    'Estimated Income',
    'Superannuation Savings',
    'Credit Card Balance',
    'Bank Loans',
    'Bank Loans',
    'Checking Accounts',
    'Saving Accounts',
    'Foreign Currency Account',
    'Business Lending'
]

# Univariate analysis and visualization
plt.figure(figsize=(8, 8))

for i, col in enumerate(numerical_cols):
    
    plt.subplot(4, 3, i + 1)
    
    sns.histplot(
        data=df,
        x=col,
        kde=True
    )
    
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")

plt.tight_layout()
plt.show()
```

# Heatmmaps



```python


numerical_cols = [
    'Estimated Income',
    'Superannuation Savings',
    'Credit Card Balance',
    'Bank Loans',
    'Checking Accounts',
    'Saving Accounts',
    'Foreign Currency Account',
    'Business Lending'
]

# Calculate correlation matrix
correlation_matrix = df[numerical_cols].corr()

# Plot heatmap
plt.figure(figsize=(12, 12))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='crest',
    fmt='.2f'
)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

## 📌 EDA Insight: Account Balance Correlation Analysis

### 🔍 Key Finding
The strongest positive correlations are observed among **Bank Deposits, Checking Accounts, Saving Accounts, and Foreign Currency Accounts**.

### 📊 Interpretation
This indicates that customers who maintain **higher balances in one type of account** are likely to maintain **substantial funds across other account types as well**. 

The positive relationship suggests that customers with stronger financial positions tend to diversify their funds across multiple banking products rather than keeping their money in a single account.

### 💡 Business Insight
- Customers with higher account balances may represent **high-value banking customers**.
- These customers can be targeted for:
  - Premium banking services
  - Investment products
  - Wealth management solutions
  - Personalized financial offers

### ✅ Conclusion
The correlation analysis highlights a pattern of **financial stability and multi-account usage behavior**, helping banks identify valuable customer segments and improve customer relationship strategies.
